```python
import numpy as np

marks = np.array([65, 72, 80, 55, 90, 68, 76, 85, 60, 95])

print("Marks:")
print(marks)

print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

print("Marks greater than 75:")
print(marks[marks > 75])
```

```python
import numpy as np

temperature = np.array([28, 31, 29, 32, 34, 30, 27])

print("Temperature:", temperature)
print("Average:", np.mean(temperature))
print("Highest:", np.max(temperature))
print("Lowest:", np.min(temperature))
print("Above 30:", temperature[temperature > 30])

temperature = temperature + 2
print("Updated Temperature:", temperature)
```

```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Chitra", "Deepak", "Esha", "Farhan", "Gokul", "Hari"],
    "Department": ["IT", "CSE", "IT", "ECE", "CSE", "IT", "ECE", "CSE"],
    "Marks": [85, 72, 90, 65, 78, 88, 55, 95],
    "Attendance": [90, 85, 95, 75, 82, 88, 70, 92]
}

df = pd.DataFrame(data)

print(df.head())
print("Average Marks:", df["Marks"].mean())

print("Marks above 75:")
print(df[df["Marks"] > 75])

print("Attendance below 80:")
print(df[df["Attendance"] < 80])

print("Sorted by Marks:")
print(df.sort_values("Marks"))
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Keyboard", "Mouse", "Headphone"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories"],
    "Price": [50000, 25000, 1500, 800, 2000],
    "Quantity": [20, 60, 70, 100, 55]
}

df = pd.DataFrame(data)

df["Total Sales"] = df["Price"] * df["Quantity"]

print(df)

print("Highest Sales:")
print(df[df["Total Sales"] == df["Total Sales"].max()])

print("Average Price:", df["Price"].mean())

print("Quantity above 50:")
print(df[df["Quantity"] > 50])

print("Sorted by Sales:")
print(df.sort_values("Total Sales"))
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 160, 200, 220]

plt.plot(months, sales, marker="o", label="Sales")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()

plt.bar(months, sales, label="Sales")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
